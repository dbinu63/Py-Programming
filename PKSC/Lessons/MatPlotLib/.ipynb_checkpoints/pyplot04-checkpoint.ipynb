# Data Visualization Demo

Using <code>matplotlib</code> we can draw the following charts:
- Line chart
- Bar chart
- Pie chart
- Histogram
- Plot chart

```python
import matplotlib.pyplot as plt
```

```python
hours = [0, 1, 2, 3, 4, 5]
temp = [45, 60, 65, 71, 67, 60]

plt.plot(hours, temp)
plt.show()
```

## <code>plot()</code> in detail

```python
hours = [0, 1, 2, 3, 4, 5]
temp = [45, 60, 65, 71, 67, 60]

plt.plot(hours, temp, color="green", marker="o", linestyle="-",linewidth=2, markersize=7,label="Temperature")
plt.show()
```

marker accepts: 'o', '*', '^', 's', 'x'\
linestyle accepts: ':', '--', '-', '-.'

```python
hours = [0, 1, 2, 3, 4, 5]
temp = [45, 60, 65, 71, 67, 60]

plt.plot(hours, temp, color="magenta", marker="o", linestyle=":",linewidth=2, markersize=7,label="Temperature")
plt.title("Temperature of the Day")
plt.xlabel("Hours")
plt.ylabel("Temperature")
plt.grid(True)
plt.legend()
plt.show()
```
