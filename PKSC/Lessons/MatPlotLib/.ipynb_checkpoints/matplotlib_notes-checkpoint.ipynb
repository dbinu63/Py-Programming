# Matplotlib Notes

```python
import matplotlib as mpl
import matplotlib.pyplot as plt
import matplotlib.colors as col
```

## 1. Basic line plot

```python
def main():
    hours = [0, 1, 2, 3, 4, 5]
    temp = [45, 60, 65, 71, 67, 60]

    plt.plot(hours, temp)
    plt.show()


if __name__ == "__main__":
    main()
```

## 2. Line styling

```python
def main():
    hours = [0, 1, 2, 3, 4, 5]
    temp = [45, 60, 65, 71, 67, 60]

    plt.plot(hours, temp, color="green", marker="o", linestyle="-",linewidth=2, markersize=7,label="Temperature")
    plt.show()

if __name__ == "__main__":
    main()
```

Marker accepts: \
'o', '*', '^', 's', 'x' 

LineStyle accepts: \
':', '--', '-', '-.'

Color accepts: \
red, green, blue, orange, yellow, coral, gold, magenta, brown, black, white, teal, lime, purple,
cyan, pink, gray, navy, tomato, olive, salmon, indigo, royalblue, skyblue, darkred, chocolate,
and hexadecimal values

## 3. Title, labels, grid, legend

```python
def main():
    hours = [0, 1, 2, 3, 4, 5]
    temp = [45, 60, 65, 71, 67, 60]

    plt.plot(hours, temp, color="magenta", marker="o", linestyle=":",linewidth=2, markersize=7,label="Temperature")
    plt.title("Temperature of the Day")
    plt.xlabel("Hours")
    plt.ylabel("Temperature")
    plt.grid(True)
    plt.legend()
    plt.show()

if __name__ == "__main__":
    main()
```

</break></break>

</break></break>

## 4. Axis limits and ticks

```python
def main(): 
    x = [1, 2, 3, 4, 5]
    y = [10, 20, 15, 25, 30]

    plt.plot(x, y, marker="o")

    plt.xlim(0, 6)
    plt.ylim(0, 35)

    plt.xticks([1, 2, 3, 4, 5])
    plt.yticks([0, 10, 20, 30, 40])

    plt.show()

if __name__ == "__main__":
    main()
```

</break></break>

</break></break>

## 5. Multiple lines

```python
def main(): 
    days = [1, 2, 3, 4, 5]
    city1 = [25, 30, 25, 27, 21]
    city2 = [32, 35, 31, 28, 30]

    plt.plot(days, city1, marker="o", color="olive", label="Houston")
    plt.plot(days, city2, marker="*", color="pink", label="Dallas")

    plt.xlim(0, 6)
    plt.ylim(10, 40)

    plt.xticks([0, 1, 2, 3, 4, 5, 6])
    plt.yticks([10, 15, 20, 25, 30, 35, 40])

    plt.title("Temperature Comparison Between Two Cities")
    plt.xlabel("Day")
    plt.ylabel("Temperature")

    plt.legend()
    plt.grid(True)

    plt.show()

if __name__ == "__main__":
    main()
```

</break></break>

</break></break>

## 6. Bar chart

```python
def main(): 
    subject = ["C", "Python", "VB", "Java"]
    marks = [85, 78, 92, 88]

    plt.bar(subject, marks, color="skyblue")
    plt.title("Marks by Subject")
    plt.xlabel("Subject")
    plt.ylabel("Marks")
    plt.ylim(0, 100)

    plt.show()

if __name__ == "__main__":
    main()
```

## 7. Bar colors: predefined names

```python
def main(): 
    subject = ["C", "Python", "VB", "Java"]
    marks = [85, 78, 92, 88]
    colors = ["darkred", "olive", "indigo", "royalblue"]

    plt.bar(subject, marks, color=colors)
    plt.title("Marks by Subject")
    plt.xlabel("Subject")
    plt.ylabel("Marks")
    plt.ylim(0, 100)

    plt.show()

if __name__ == "__main__":
    main()
```

You can pass lists of colors where appropriate.

</break></break>

</break></break>

## 8. Bar colors: gradient

```python
def main(): 
    subject = ["C", "Python", "VB", "Java"]
    marks = [85, 78, 92, 88]
    colors = ["darkred", "olive", "indigo", "royalblue"]
    gradient = []

    for color in colors: 
        cmap = col.LinearSegmentedColormap.from_list("gradient", ["white", color])
        gradient.append(cmap(0.5))

    plt.bar(subject, marks, color=gradient)
    plt.title("Marks by Subject")
    plt.xlabel("Subject")
    plt.ylabel("Marks")
    plt.ylim(0, 100)

    plt.show()

if __name__ == "__main__":
    main()
```

This snippet can be a little system-specific. 

The 0.5 in "gradient.append(cmap(0.5))" sets the color halfway across the selected gradient. This value is not optional. 

</break></break>

</break></break>

## 9. Bar colors: hex

```python
def main(): 
    subject = ["C", "Python", "VB", "Java"]
    marks = [85, 78, 92, 88]

    plt.bar(subject, marks, color=["#29C287", "#DE9E14", "#5414DE", "#DE1439"])
    plt.title("Marks by Subject")
    plt.xlabel("Subject")
    plt.ylabel("Marks")
    plt.ylim(0, 100)

    plt.show()

if __name__ == "__main__":
    main()
```

Python doesn't work with the typical 0-255 values for RGB. Rather, it works with decimal values 0-1. (For example, R(200), G(50), B(250) should be written as (in decimals, set in a tuple) 200/255, 50/255, 250/255). 

Alternatively, you can simply use hexadecimal values.

## 10. Scatter plot

```python
def main(): 
    hours_studied = [1, 2, 3, 4, 5, 6, 7]
    marks = [25, 40, 50, 62, 78, 85, 93]

    plt.scatter(hours_studied, marks, color="purple", alpha=1, s=50)

    plt.title("Hours Studied v Marks")
    plt.xlabel("Hours Studied")
    plt.ylabel("Marks")

    plt.show()

if __name__ == "__main__":
    main()
```

Alpha is the transparency key. S sets the size of the scatter points. 

</break></break>

</break></break>

## 11. Pie chart

```python
def main(): 
    subject = ["C", "Python", "VB", "Java"]
    marks = [30, 40, 20, 10]

    plt.pie(marks, labels=subject, colors=["red", "green", "blue", "yellow"], startangle=90, autopct="%1.1f%%")
    # red slice will start at 90 degree north of east
    
    plt.title("Marks Per Subject") 
    plt.show()

if __name__ == "__main__":
    main()
```

The startangle sets the position the first slice will start from, in terms of (x) degrees north of east. In this example, the first (red) slice will start 90 degrees north of east, or at the vertical edge of the second quadrant. 

</break></break>

</break></break>

## 12. Histogram

```python
def main(): 
    marks = [83, 27, 94, 56, 72, 83, 19, 67, 94, 41, 100, 56, 34, 87, 12, 72, 98, 46, 67, 90]

    plt.hist(marks, bins=5, color="red", edgecolor="black")

    plt.title("Distribution of Exam Marks")
    plt.xlabel("Marks")
    plt.ylabel("Number of Students")

    plt.show()

if __name__ == "__main__":
    main()
```
